```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("Unemployment in India.csv")
```

#Data Cleaning

```python

df.columns = df.columns.str.strip()

# Convert date
df['Date'] = pd.to_datetime(df['Date'])

# Handle missing values
df = df.dropna()

# Extract month & year
df['Month'] = df['Date'].dt.month
df['Year'] = df['Date'].dt.year
```

#EDA

```python

print(df.info())
print(df.describe())

# Unemployment distribution
sns.histplot(df['Estimated Unemployment Rate (%)'], kde=True)
plt.show()
```

#Trend Analysis

```python
plt.figure(figsize=(10,5))
df.groupby('Date')['Estimated Unemployment Rate (%)'].mean().plot()
plt.title("Unemployment Trend Over Time")
plt.show()
```

#COVID-19 Impact Analysis

```python
covid = df[df['Year'] == 2020]

sns.lineplot(data=covid, x='Date', y='Estimated Unemployment Rate (%)')
plt.title("Unemployment During COVID-19")
plt.show()
```

#State-wise Analysis

```python
top_states = df.groupby('Region')['Estimated Unemployment Rate (%)'].mean().sort_values(ascending=False)

top_states.head(10).plot(kind='bar')
plt.title("Top Unemployment States")
plt.show()
```

#Seasonal Trends

```python
sns.boxplot(x='Month', y='Estimated Unemployment Rate (%)', data=df)
plt.title("Seasonal Unemployment Trends")
plt.show()
```

#Correlation Analysis

```python
sns.heatmap(df.select_dtypes(include=['number']).corr(), annot=True, cmap='coolwarm')
plt.show()
```

#Predict Unemployment Rate

```python
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestRegressor

X = df[['Estimated Employed', 'Estimated Labour Participation Rate (%)']]
y = df['Estimated Unemployment Rate (%)']

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2)

model = RandomForestRegressor()
model.fit(X_train, y_train)

print("Accuracy:", model.score(X_test, y_test))
```
